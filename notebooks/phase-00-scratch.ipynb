```python
import os

data = load_dataset("stanfordnlp/imdb", split="train")
data.to_csv("imdb.csv")
data.to_parquet("imdb.parquet")

for f in ["imdb.csv", "imdb.parquet"]:
    print(f, f"{os.path.getsize(f)/1e6:.1f} MB")
```

```python
data = load_dataset("stanfordnlp/imdb", split="train")

split = data.train_test_split(test_size=0.3, seed=42)
temp = split["test"].train_test_split(test_size=0.5, seed=42)

train_ds, val_ds, test_ds = split["train"], temp["train"], temp["test"]
print(f"Train: {len(train_ds)}, Val: {len(val_ds)}, Test: {len(test_ds)}")
```

```python
print(Counter(train_ds["label"]), Counter(test_ds["label"]))
```

```python
stream = load_dataset("stanfordnlp/imdb", split="train", streaming=True)

for i, example in enumerate(stream):
    print(i, example["label"], example["text"][:60])
    if i >= 4:
        break
```
